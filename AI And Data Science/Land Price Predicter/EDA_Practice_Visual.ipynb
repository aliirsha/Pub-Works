```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('landprices.csv')
```

```python
print(df.head())
```

```python
df_location_prices = df[['location_name', 'price']]
```

```python
print(df_location_prices)
```

```python
df_size_prices = df[['land_area_cents', 'price']]
```

```python
print(df_size_prices.describe())
```

```python
df['price_per_cent'] = df['price'] / df['land_area_cents']
print(df.head())
```

```python
print(df['price_per_cent'].describe())
```

```python
sns.histplot(data=df, x='price_per_cent')
plt.title('Ditribution of price')
plt.show()
```

```python
sns.scatterplot(data=df, x='land_area_cents', y='price_per_cent')
plt.title('Land size vs price')
plt.show()
```

```python
sns.boxplot(data=df, x='land_type', y='price_per_cent' )
plt.title('land type vs price')
plt.show()
```

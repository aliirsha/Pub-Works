### Master Calendar Generator
This script creates a robust master calendar with common time dimensions (Year, Quarter, Month, Week, etc.) and saves it to an 'Output' folder.

```python
import pandas as pd
import os
from datetime import date
import ipywidgets as widgets
from IPython.display import display

# Create Output directory if it doesn't exist
output_dir = 'Output'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"Created directory: {output_dir}")

def generate_master_calendar(start_date, end_date):
    # Generate date range
    dates = pd.date_range(start=start_date, end=end_date)
    
    # Create DataFrame
    df = pd.DataFrame({'Date': dates})
    
    # Optimized column generation
    df['Year'] = df['Date'].dt.year
    df['Quarter'] = df['Date'].dt.quarter
    df['Month'] = df['Date'].dt.month
    df['MonthName'] = df['Date'].dt.month_name()
    df['WeekNumber'] = df['Date'].dt.isocalendar().week
    df['Day'] = df['Date'].dt.day
    df['DayOfWeek'] = df['Date'].dt.dayofweek
    df['DayName'] = df['Date'].dt.day_name()
    df['IsWeekend'] = df['DayOfWeek'].isin([5, 6])
    df['YearMonth'] = df['Date'].dt.to_period('M')
    
    # Save to CSV
    file_path = os.path.join(output_dir, 'master_calendar.csv')
    df.to_csv(file_path, index=False)
    print(f"Master calendar saved successfully to: {file_path}")
    return df

# UI components for Date Input
start_picker = widgets.DatePicker(description='From Date', value=date(2023, 1, 1))
end_picker = widgets.DatePicker(description='To Date', value=date(2023, 12, 31))
btn = widgets.Button(description="Generate Calendar", button_style='success')

def on_button_clicked(b):
    if start_picker.value and end_picker.value:
        if start_picker.value <= end_picker.value:
            calendar_df = generate_master_calendar(start_picker.value, end_picker.value)
            display(calendar_df.head())
        else:
            print("Error: 'From Date' must be before 'To Date'")
    else:
        print("Please select both dates.")

btn.on_click(on_button_clicked)
display(widgets.VBox([start_picker, end_picker, btn]))
```
